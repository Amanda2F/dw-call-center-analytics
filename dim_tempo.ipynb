```python
import pandas as pd
```

```python
datas = pd.date_range(start='2026-08-12', end='2026-09-12')
```

```python
df = pd.DataFrame({'data_completa': datas})
```

```python
df['id_data'] = df['data_completa'].dt.strftime('%Y%m%d').astype(int)
df['ano'] = df['data_completa'].dt.year
df['mes'] = df['data_completa'].dt.month
df['dia'] = df['data_completa'].dt.day
```

```python
dias_semana_pt = {
    0: 'Segunda-feira', 1: 'Terça-feira', 2: 'Quarta-feira',
    3: 'Quinta-feira', 4: 'Sexta-feira', 5: 'Sábado', 6: 'Domingo'
}
df['dia_semana'] = df['data_completa'].dt.dayofweek.map(dias_semana_pt)
df['eh_fim_semana'] = df['data_completa'].dt.dayofweek.isin([5, 6])
```

```python
df = df[['id_data', 'data_completa', 'ano', 'mes', 'dia', 'dia_semana', 'eh_fim_semana']]
```

```python
salvar_csv = 'Arquivos CSV/dim_tempo.csv'
```

```python
df.to_csv(salvar_csv, index=False)
```
