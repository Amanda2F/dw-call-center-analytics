```python
import pandas as pd
```

```python
colunas_origem = ['qualification_name','qualification_behavior_text','list_id','list_name']
```

```python
caminho_leitura = 'Arquivos CSV/Diario/01.09_a_23.09_ligacoes_detalhadas.csv'
df = pd.read_csv(caminho_leitura, usecols= colunas_origem, sep=';')
```

```python
df = df.rename(columns={
    'qualification_name': 'qualificacao',
    'qualification_behavior_text': 'comportamento',
    'list_id': 'lista_id',
    'list_name': 'lista_nome'
})
```

```python
traducao_comportamento = {
    'not-call-identifier': 'sem-identificador-de-chamada',
    'not-call-phone': 'não-ligar-telefone',
    'repeat': 'repetir'
}

df['comportamento'] = df['comportamento'].map(traducao_comportamento).fillna(df['comportamento'])
```

```python
df['lista_id'] = df['lista_id'].fillna(-1).astype(int)
df['lista_nome'] = df['lista_nome'].fillna('Não Informado')
df['qualificacao'] = df['qualificacao'].fillna('Não Informado')
df['comportamento'] = df['comportamento'].fillna('Não Informado')
```

```python
df = df.drop_duplicates().reset_index(drop=True)
```

```python
salvar_csv = 'Arquivos CSV/dim_caracteristicas.csv'
```

```python
df.to_csv(salvar_csv, index=False)
```
