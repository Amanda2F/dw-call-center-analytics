```python
import pandas as pd
```

```python
caminho_leitura = 'Arquivos CSV/Diario/01.09_a_23.09_ligacoes_detalhadas.csv'
df = pd.read_csv(caminho_leitura, usecols=['agent_name'], sep=';')
```

```python
df = df.drop_duplicates().dropna()
```

```python
df = df.rename(columns={
    'agent_name': 'nome_agente'
})
```

```python
df = df.sample(frac=1).reset_index(drop=True)
```

```python
df['nome_oculto'] = [f"Operador {i+1}" for i in range(len(df))]
```

```python
salvar_csv = 'Arquivos CSV/dim_agentes_confidencial.csv'
```

```python
df.to_csv(salvar_csv, index=False)
```
