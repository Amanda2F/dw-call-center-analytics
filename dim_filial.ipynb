```python
import pandas as pd
```

```python
caminho_leitura = 'Arquivos CSV/Diario/12.08_a_12.09_ligacoes_detalhadas.csv'
df = pd.read_csv(caminho_leitura, usecols=['campaign_name'])
```

```python
df = df.drop_duplicates().dropna()
```

```python
df[['sigla_uf', 'loja']] = df['campaign_name'].str.split(' ', n=1, expand=True)
```

```python
df = df[['sigla_uf', 'loja']].drop_duplicates()
```

```python
df = df.sample(frac=1).reset_index(drop=True)
```

```python
df['nome_oculto'] = [f"Filial {i+1}" for i in range(len(df))]
```

```python
salvar_csv = 'Arquivos CSV/dim_filial_confidencial.csv'
```

```python
df.to_csv(salvar_csv, index=False)
```
